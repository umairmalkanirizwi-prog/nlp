```python
print("Name:Umair Malkani. UIN:242A004")
documents = [

  "Artificial intelligence allows machines to perform tasks using learned information.",
  "Data analysis helps computers identify useful trends and relationships.",
  "Speech recognition enables software to process and interpret spoken words.",
  "The students completed their project before the deadline.",
  "Neural networks are commonly used in modern artificial intelligence applications."
]
```

```python
print("Name:Umair Malkani. UIN:242A004")
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import pandas as pd
```

```python
for i, document in enumerate(documents):
    print(f"D{i+1}: {document}")
```

```python
print("Name:Umair Malkani. UIN:242A004")
vectorizer = TfidfVectorizer(stop_words='english')
tfidf_matrix = vectorizer.fit_transform(documents)
print(tfidf_matrix.shape)
```

```python
print("Name:Umair Malkani. UIN:242A004")
tfidf_df = pd.DataFrame(
tfidf_matrix.toarray(),
columns=vectorizer.get_feature_names_out(),
index=["D1", "D2", "D3", "D4", "D5"]
)
tfidf_df.round(2)
```

```python
print("Name:Umair Malkani. UIN:242A004")
print(tfidf_matrix)
```

```python
print("Name:Umair Malkani. UIN:242A004")
print(vectorizer.get_feature_names_out())
```

```python
print("Name:Umair Malkani. UIN:242A004")
similarity_matrix = cosine_similarity(tfidf_matrix)
similarity_matrix
```

```python
similarity_df = pd.DataFrame (
similarity_matrix,
index=["D1", "D2", "D3", "D4", "D5"],
columns=["D1", "D2", "D3", "D4", "D5"]
)
similarity_df.round(3)
```

```python
print("Name:Umair Malkani. UIN:242A004")
highest_similarity = -1
most_similar_pair = None
for i in range(len(documents)):
    for j in range(i + 1, len(documents)): # Corrected range syntax and indentation
        similarity = similarity_matrix[i][j]
        if similarity > highest_similarity:
            highest_similarity = similarity
            most_similar_pair = (i, j) # Corrected to assignment

if most_similar_pair:
    print(
        f"Most similar documents:\n" # Corrected f-string
        f"D{most_similar_pair[0]+1} and D{most_similar_pair[1]+1}"
    )
    print(f"Similarity: {highest_similarity:.3f}") # Corrected f-string
else:
    print("No similar pairs found.")
```

```python
print("Name:Umair Malkani. UIN:242A004")
def similar (D1, D5):
    vector = vectorizer.fit_transform([D1,D5])
    similarity=cosine_similarity(vector)
    return similarity
```
