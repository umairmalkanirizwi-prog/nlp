```python
print("Umair Malkani")
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import UnigramTagger, BigramTagger, pos_tag
from nltk.corpus import treebank
from nltk import pos_tag_sents
```

```python
print("Umair Malkani")
# Download required NLTK data files
nltk.download('averaged_perceptron_tagger')
nltk.download('punkt')
nltk.download('treebank')
```

```python
print("Umair Malkani")
# Sample text for tagging
sample_text = "The quick brown fox jumps over the lazy dog. The lazy dog barks."

print(sample_text)
```

```python
print("Umair Malkani")
nltk.download('punkt_tab')

# Tokenize the text
tokens = word_tokenize(sample_text)

print(tokens)
```

```python
print("Umair Malkani")
nltk.download('averaged_perceptron_tagger_eng')

# POS Tagging using NLTK's built-in statistical tagger
rule_based_tags = pos_tag(tokens)

print(rule_based_tags)
```

```python
print("Umair Malkani")
nltk.download('tagsets_json')
nltk.help.upenn_tagset("NN")
```

```python
print("Umair Malkani")
# Statistical Tagging using Treebank corpus-trained tagger

# Get tagged sentences from Treebank corpus
train_sents = treebank.tagged_sents()

# Create and train the Unigram Tagger
unigram_tagger = UnigramTagger(train_sents)

# Tag the sample tokens
unigram_tags = unigram_tagger.tag(tokens)

# Display the result
print(unigram_tags)
```

```python
print("Umair Malkani")
# Bigram Tagger using Treebank corpus

bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)

# Tag the sample tokens
bigram_tags = bigram_tagger.tag(tokens)

print(bigram_tags)
```

```python
print("Umair Malkani")
# Compare Rule-based/Statistical, Unigram and Bigram Tagging

# 1. NLTK built-in POS tagger
rule_based_tags = pos_tag(tokens)

# 2. Train Unigram Tagger using Treebank corpus
train_sents = treebank.tagged_sents()
unigram_tagger = UnigramTagger(train_sents)
unigram_tags = unigram_tagger.tag(tokens)

# 3. Train Bigram Tagger using Unigram as backoff
bigram_tagger = BigramTagger(train_sents, backoff=unigram_tagger)
bigram_tags = bigram_tagger.tag(tokens)

# Print results
print("WORD".ljust(12), "NLTK TAG".ljust(12), "UNIGRAM".ljust(12), "BIGRAM")
print("-" * 50)

for i in range(len(tokens)):
    word = tokens[i]
    nltk_tag = rule_based_tags[i][1]
    unigram_tag = unigram_tags[i][1]
    bigram_tag = bigram_tags[i][1]

    print(
        word.ljust(12),
        str(nltk_tag).ljust(12),
        str(unigram_tag).ljust(12),
        str(bigram_tag)
    )
```

```python
print("Umair Malkani")
# Display the results for comparison

print("Original Sentence:", sample_text)

print("\nRule-Based POS Tagging (Statistical):")
print(rule_based_tags)

print("\nUnigram POS Tagging:")
print(unigram_tags)

print("\nBigram POS Tagging:")
print(bigram_tags)
```

```python
# Evaluate the performance of each tagger using Treebank
test_sents=treebank.tagged_sents()[3000:]
unigram_acc=unigram_tagger.evaluate(test_sents)
bigram_acc = bigram_tagger.evaluate(test_sents)
print(f"\nUnigram Tagger Accuracy: {unigram_acc * 100:.2f}%")
print(f"Bigram Tagger Accuracy: {bigram_acc * 100:.2f}%")
```
