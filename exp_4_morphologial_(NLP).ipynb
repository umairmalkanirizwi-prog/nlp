```python
print("umair Malkani  242004")
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
print("umair Malkani  242004")
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
print("umair Malkani  242004")
import string
print(string.punctuation)
```

```python
print("umair Malkani  242004")
text_lower=text.lower()
print(text_lower)
```

```python
print("umair Malkani  242004")
tokens = word_tokenize(text_lower)
print(tokens)
```

```python
print("umair Malkani  242004")
text_clean=[]
for word in tokens:
    if word not in string.punctuation:
        text_clean.append(word)
print(text_clean)
```

```python
print("umair Malkani  242004")
lematizer = WordNetLemmatizer()
lemmas = [lematizer.lemmatize(tok, pos='v') for tok in text_clean]
lemmas
```

```python
print("umair Malkani  242004")
lematizer.lemmatize("computer",pos='n')
```

```python
print("umair Malkani  242004")
[(token, lemma) for token, lemma in zip(tokens, lemmas)]
```

```python
print("umair Malkani  242004")
morphological_info = {}
for token, lemma in zip(tokens, lemmas):
    if token != lemma:
        if token.endswith('ing'):
            analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
        elif token.endswith("er") or token.endswith("ation"):
            analysis = "Derivational: '{}' + suffix indicating a derived form".format(lemma)
        else:
            analysis = "Inflectional or irregular form of '{}'".format(lemma)
    else:
        analysis = "Base form of '{}'".format(lemma)

    morphological_info[token] = analysis
```

```python
print("umair Malkani  242004")
morphological_info
```
