```python
print("umair malkani", "UIN: 242A004")
import nltk
nltk.download('punkt_tab')
```

```python
print("umair malkani", "UIN: 242A004")
from nltk import word_tokenize
tokenized=word_tokenize(text_emoji)
print("tokenized words",tokenized)
```

```python
print("umair malkani", "UIN: 242A004")
import string
punctuation=list(string.punctuation)
print(punctuation)
```

```python
print("umair malkani", "UIN: 242A004")
print("With punctuation:", tokenized)
print()

text_punct = []

for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)

print("Without punctuation:", text_punct)
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
print((stopwords.words('english')))
```

```python
text_clean2=[]
for word in text_punct:
  if word not in stopwords.words('english'):
    text_clean2.append(word)
print(text_clean2)
```

```python
from nltk.stem import PorterStemmer
stemmer=PorterStemmer()
print([stemmer.stem(word)for word in text_clean2])
```
