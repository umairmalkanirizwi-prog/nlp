```python
def preprocess_tweet(tweet_text):
    # Remove old style retweet text "RT"
    tweet_cleaned = re.sub(r'^RT[\s]+', '', tweet_text)

    # Remove hyperlinks
    tweet_cleaned = re.sub(r'https?://[^\s\n\r]+', '', tweet_cleaned)

    # Remove hashtags and mentions (by removing the # and @ signs)
    tweet_cleaned = re.sub(r'#', '', tweet_cleaned)
    tweet_cleaned = re.sub(r'@', '', tweet_cleaned)

    # Remove specific punctuation identified earlier (emoticons, etc.)
    tweet_cleaned = re.sub(r'[():]', '', tweet_cleaned)

    # Tokenize the tweet
    tokenizer_func = TweetTokenizer(preserve_case=False, strip_handles=True, reduce_len=True)
    tweet_tokens = tokenizer_func.tokenize(tweet_cleaned)

    # Remove the first word if it exists (as per previous user request)
    if len(tweet_tokens) > 0:
        tweet_tokens = tweet_tokens[1:]

    # Remove stopwords and punctuation
    tweets_clean_final = []
    for word in tweet_tokens:
        if (word not in stopwords_english and
                word not in string.punctuation):
            tweets_clean_final.append(word)

    return tweets_clean_final
```

```python
all_tweets_df['processed_tweets'] = all_tweets_df['tweets'].apply(preprocess_tweet)
display(all_tweets_df[['tweets', 'processed_tweets']].head())
```
